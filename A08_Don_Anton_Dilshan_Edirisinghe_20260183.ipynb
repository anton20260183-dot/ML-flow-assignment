Goal: Use a pretrained model (e.g., distilbert-base-uncased-finetuned-sst-2-english)

• Use Hugging Face pipeline or Trainer API

• Log the following in MLflow:– Model name– Number of samples– Accuracy– Predictions as an artifact (CSV)

Task 1 run 1

```python
import os

os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from transformers import pipeline
import mlflow
import pandas as pd

# Pretrained model
model_1 = "distilbert-base-uncased-finetuned-sst-2-english"

sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model=model_1,
    device=-1
)

print("Model loaded successfully!")
```

```python
sentences_run1 = [
    "The new laptop works incredibly well.",
    "I was impressed by how quickly the package arrived.",
    "The staff were friendly and helpful.",
    "The hotel provided excellent service throughout my stay.",
    "I am pleased with the quality of this product.",
    "The presentation was clear, informative, and engaging.",
    "The app makes managing my tasks much easier.",
    "The meal was delicious and beautifully prepared.",
    "The training session was useful and well organized.",
    "I had a pleasant experience with the company.",

    "The laptop stopped working after only a few days.",
    "The package arrived late and damaged.",
    "The staff were rude and unhelpful.",
    "The hotel service was disappointing.",
    "The product feels cheap and poorly made.",
    "The presentation was confusing and difficult to follow.",
    "The app is slow and frustrating to use.",
    "The meal was cold and tasteless.",
    "The training session was poorly organized and unhelpful.",
    "I had a terrible experience with the company."
]

true_labels_run1 = [
    "POSITIVE", "POSITIVE", "POSITIVE", "POSITIVE", "POSITIVE",
    "POSITIVE", "POSITIVE", "POSITIVE", "POSITIVE", "POSITIVE",

    "NEGATIVE", "NEGATIVE", "NEGATIVE", "NEGATIVE", "NEGATIVE",
    "NEGATIVE", "NEGATIVE", "NEGATIVE", "NEGATIVE", "NEGATIVE"
]

print("Run 1 samples:", len(sentences_run1))
```

```python
predicted_labels_run1 = []
confidence_scores_run1 = []

print("Starting Run 1 inference...\n")

for i, sentence in enumerate(sentences_run1):

    result = sentiment_pipeline(sentence)[0]

    predicted_labels_run1.append(result["label"])
    confidence_scores_run1.append(result["score"])

    print(f"{i + 1}/20 completed")

print("\nRun 1 inference completed!")
```

```python
correct_run1 = sum(
    actual == predicted
    for actual, predicted in zip(
        true_labels_run1,
        predicted_labels_run1
    )
)

accuracy_run1 = correct_run1 / len(sentences_run1)

print("Run 1 Accuracy:", accuracy_run1)

results_run1 = pd.DataFrame({
    "sentence": sentences_run1,
    "true_label": true_labels_run1,
    "predicted_label": predicted_labels_run1,
    "confidence": confidence_scores_run1
})

csv_run1 = "sentiment_predictions_run1.csv"

results_run1.to_csv(csv_run1, index=False)

display(results_run1)
```

```python
mlflow.set_experiment("HuggingFace_Sentiment_Analysis")

with mlflow.start_run(run_name="Sentiment_model_Run_1"):

    mlflow.log_param("model_name", model_1)
    mlflow.log_param("number_of_samples", len(sentences_run1))

    mlflow.log_metric("accuracy", accuracy_run1)

    mlflow.log_artifact(csv_run1)

    print("MLflow for sentiment analysis Run 1 completed!")
```

Task 1 run 1

```python
import os

os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from transformers import pipeline
import mlflow
import pandas as pd
from collections import Counter

# Pretrained NER model
ner_model_name = "dslim/bert-base-NER"

ner_pipeline = pipeline(
    "ner",
    model=ner_model_name,
    aggregation_strategy="simple",
    device=-1
)

print("NER model loaded successfully!")
```

```python
paragraphs_run1 = [
    """Mary Kom trained for many years in Manipur before becoming an
    internationally recognized boxer. She represented India in several
    competitions and received support from the Sports Authority of India
    in New Delhi.""",

    """Tim Berners-Lee invented the World Wide Web while working at CERN
    in Geneva. His work changed how people access information on the
    internet, and CERN continues to support scientific research in
    Switzerland.""",

    """Indra Nooyi served as the chief executive officer of PepsiCo.
    During her career, she worked with teams in New York and other parts
    of the United States to expand the company's global business.""",

    """The World Health Organization organized a conference in Geneva
    where Tedros Adhanom Ghebreyesus spoke about international health
    cooperation. Representatives from Japan, Canada, and Australia
    attended the event.""",

    """Sachin Tendulkar played international cricket for India for many
    years. He represented Mumbai in domestic competitions and played at
    famous stadiums across India, including venues in Mumbai and Delhi.""",

    """Amazon Web Services announced a new cloud computing program in
    Austin. Andy Jassy discussed the company's plans to help businesses
    across Texas adopt cloud technologies and improve their digital
    infrastructure.""",

    """Malala Yousafzai visited New York to attend an education event
    organized by the Malala Fund. She spoke with students and educators
    about improving access to education for children around the world.""",

    """Toyota introduced a new electric vehicle at an event in Tokyo.
    Akio Toyoda discussed the company's plans to invest in sustainable
    transportation and expand production facilities across Japan.""",

    """Leonardo da Vinci spent part of his life in Florence, where he
    created several important works of art. His work was later studied
    by researchers and historians at institutions throughout Italy.""",

    """The International Monetary Fund held discussions in Washington,
    D.C., with representatives from Germany, France, and Brazil.
    Kristalina Georgieva addressed economic challenges and encouraged
    cooperation between member countries."""
]

print("Run 1 paragraphs:", len(paragraphs_run1))
```

```python
entities_run1 = []

print("Starting Run 1 NER...\n")

for i, paragraph in enumerate(paragraphs_run1):

    results = ner_pipeline(paragraph)

    for entity in results:

        entity_type = entity["entity_group"]

        # Keep only required entity types
        if entity_type in ["PER", "ORG", "LOC"]:

            entities_run1.append({
                "paragraph": i + 1,
                "entity": entity["word"],
                "entity_type": entity_type,
                "confidence": entity["score"]
            })

    print(f"{i + 1}/10 paragraphs completed")

print("\nRun 1 NER completed!")
print("Total entities:", len(entities_run1))
```

```python
from collections import Counter

# Count entity types
entity_counts_run1 = Counter(
    entity["entity_type"]
    for entity in entities_run1
)

person_count_run1 = entity_counts_run1.get("PER", 0)
organization_count_run1 = entity_counts_run1.get("ORG", 0)
location_count_run1 = entity_counts_run1.get("LOC", 0)

print("Run 1 entity distribution:")
print("Person:", person_count_run1)
print("Organization:", organization_count_run1)
print("Location:", location_count_run1)
print("Total:", len(entities_run1))

# Create DataFrame
entities_df_run1 = pd.DataFrame(entities_run1)

# Save CSV
ner_csv_run1 = "ner_entities_run1.csv"

entities_df_run1.to_csv(
    ner_csv_run1,
    index=False
)

display(entities_df_run1)
```

```python
mlflow.set_experiment("HuggingFace_NER")

with mlflow.start_run(run_name="NER_Run_1"):

    # Parameter
    mlflow.log_param(
        "model_name",
        ner_model_name
    )

    # Metrics
    mlflow.log_metric(
        "total_entities_extracted",
        len(entities_run1)
    )

    mlflow.log_metric(
        "person_entities",
        person_count_run1
    )

    mlflow.log_metric(
        "organization_entities",
        organization_count_run1
    )

    mlflow.log_metric(
        "location_entities",
        location_count_run1
    )

    # Artifact
    mlflow.log_artifact(ner_csv_run1)

    print("MLflow NER Run 1 completed!")
```

Task 3: Text Generation Goal: Use a generative model (e.g., gpt2) Requirements: • Provide 5 prompts • Generate text using different parameters: – temperature – max_length • Log: – Parameters used – Generated outputs – Observations (as text artifact)


Task 3 run 1

```python
import os

os.environ["HF_HUB_DISABLE_PROGRESS_BARS"] = "1"
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
os.environ["TOKENIZERS_PARALLELISM"] = "false"

from transformers import pipeline
import mlflow
import pandas as pd

# Pretrained generative model
generation_model_name = "gpt2"

generator = pipeline(
    "text-generation",
    model=generation_model_name,
    device=-1
)

print("GPT-2 model loaded successfully!")
```

```python
prompts_run1 = [
    "A healthy lifestyle begins with",
    "The best way to solve a difficult problem is",
    "Reading books can help people",
    "A successful team should always",
    "In the coming years, education will"
]

settings_run1 = [
    {"temperature": 0.5, "max_length": 50},
    {"temperature": 0.6, "max_length": 60},
    {"temperature": 0.5, "max_length": 55},
    {"temperature": 0.7, "max_length": 65},
    {"temperature": 0.6, "max_length": 60}
]

print("Run 1 prompts:", len(prompts_run1))
```

```python
generated_results_run1 = []

print("Starting Run 1 text generation...\n")

for i, prompt in enumerate(prompts_run1):

    settings = settings_run1[i]

    result = generator(
        prompt,
        max_length=settings["max_length"],
        temperature=settings["temperature"],
        do_sample=True,
        num_return_sequences=1
    )

    generated_text = result[0]["generated_text"]

    generated_results_run1.append({
        "prompt": prompt,
        "temperature": settings["temperature"],
        "max_length": settings["max_length"],
        "generated_text": generated_text
    })

    print(f"{i + 1}/5 completed")
    print("Generated:", generated_text)
    print("-" * 70)

print("\nRun 1 generation completed!")
```

```python
results_df_run1 = pd.DataFrame(generated_results_run1)

generation_csv_run1 = "generated_outputs_run1.csv"

results_df_run1.to_csv(
    generation_csv_run1,
    index=False
)

print("Run 1 output file created:", generation_csv_run1)

display(results_df_run1)
```

```python
observations_run1 = """
Task 3 - Text Generation Run 1 Observations

In this run, GPT-2 generated responses for five different prompts
using relatively low temperature values.

The lower temperature settings made the generated text more
consistent and focused. The outputs generally followed the context
provided by each prompt with less variation.

The max_length values determined how long the generated responses
could become. Increasing max_length allowed the model to continue
the generated text for a longer sequence.

Overall, the first run demonstrated how conservative temperature
settings can produce more stable and predictable text generation.
"""

observations_path_run1 = "generation_observations_run1.txt"

with open(
    observations_path_run1,
    "w",
    encoding="utf-8"
) as file:
    file.write(observations_run1)

print("Observations file created:", observations_path_run1)
```

Task 3 run 2

```python
prompts_run2 = [
    "Exploring a new country can teach us",
    "Creativity becomes stronger when",
    "A surprising discovery in science could",
    "Working with people from different cultures can",
    "The future of transportation might"
]

settings_run2 = [
    {"temperature": 1.0, "max_length": 70},
    {"temperature": 1.1, "max_length": 80},
    {"temperature": 1.0, "max_length": 75},
    {"temperature": 1.2, "max_length": 85},
    {"temperature": 1.1, "max_length": 75}
]

print("Run 2 prompts:", len(prompts_run2))
```

```python
generated_results_run2 = []

print("Starting Run 2 text generation...\n")

for i, prompt in enumerate(prompts_run2):

    settings = settings_run2[i]

    result = generator(
        prompt,
        max_length=settings["max_length"],
        temperature=settings["temperature"],
        do_sample=True,
        num_return_sequences=1
    )

    generated_text = result[0]["generated_text"]

    generated_results_run2.append({
        "prompt": prompt,
        "temperature": settings["temperature"],
        "max_length": settings["max_length"],
        "generated_text": generated_text
    })

    print(f"{i + 1}/5 completed")
    print("Generated:", generated_text)
    print("-" * 70)

print("\nRun 2 generation completed!")
```

```python
results_df_run2 = pd.DataFrame(generated_results_run2)

generation_csv_run2 = "generated_outputs_run2.csv"

results_df_run2.to_csv(
    generation_csv_run2,
    index=False
)

print("Run 2 output file created:", generation_csv_run2)

display(results_df_run2)
```

```python
observations_run2 = """
Task 3 - Text Generation Run 2 Observations

For the second run, GPT-2 was tested with higher temperature
settings between 1.0 and 1.2 across five different prompts.

The increased temperature allowed the model to produce more
diverse and less predictable responses. Some outputs showed
greater variation in wording and ideas compared with the first run.

The different max_length values also affected how much text the
model was able to generate for each prompt.

Overall, this run demonstrated that increasing the temperature
can encourage more varied text generation, although very high
values may sometimes reduce the consistency of the output.
"""

observations_path_run2 = "generation_observations_run2.txt"

with open(
    observations_path_run2,
    "w",
    encoding="utf-8"
) as file:
    file.write(observations_run2)

print("Observations file created:", observations_path_run2)
```

```python
mlflow.set_experiment("HuggingFace_Text_Generation")

with mlflow.start_run(run_name="Text_Generation_Run_2"):

    # Parameters
    mlflow.log_param(
        "model_name",
        generation_model_name
    )

    for i, settings in enumerate(settings_run2, start=1):

        mlflow.log_param(
            f"prompt_{i}_temperature",
            settings["temperature"]
        )

        mlflow.log_param(
            f"prompt_{i}_max_length",
            settings["max_length"]
        )

    # Metric
    mlflow.log_metric(
        "number_of_prompts",
        len(prompts_run2)
    )

    # Artifacts
    mlflow.log_artifact(generation_csv_run2)
    mlflow.log_artifact(observations_path_run2)

    print("MLflow Text Generation Run 2 completed!")
```

Observation: GPT-2 successfully generated text for all five prompts using different temperature and maximum-length settings. The lower-temperature run produced more consistent outputs, while the higher-temperature run generated more varied responses. The results demonstrate that temperature influences the creativity and randomness of generated text, whereas max_length controls the possible length of the output.
